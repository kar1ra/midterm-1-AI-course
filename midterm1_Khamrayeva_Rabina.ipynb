```python
import pandas as pd
from IPython.display import Markdown, display

display(Markdown(open("my_data/DATA.md", encoding="utf-8").read()))
ranks = pd.read_csv("my_data/retrieval_ranks.csv")   # Part 1
docs = pd.read_csv("my_data/doc_results.csv")        # Part 2
display(ranks.head(), docs)
```

```python
import math
print(ranks.columns.tolist())

def wilson(h, n, z=1.96):
    p = h/n; d = 1 + z*z/n
    c = (p + z*z/(2*n))/d
    w = z*math.sqrt(p*(1-p)/n + z*z/(4*n*n))/d
    return round(c-w,3), round(c+w,3)

for cfg in ["rank_A","rank_B"]:
    r = ranks[cfg]
    print(cfg, {f"R@{k}": round((r<=k).sum()/60,3) for k in (1,3,5)},
          "MRR@10", round((1/r).where(r<=10,0).sum()/60,3))
    for k in (3,5):
        h = (r<=k).sum(); print(" k",k,"hits",h, wilson(h,60))

lc = [c for c in ranks.columns if "lang" in c.lower()][0]
for cfg,k in [("rank_B",3),("rank_B",5)]:
    print(cfg,k, ranks.groupby(lc)[cfg].apply(lambda r:(r<=k).sum()/20))
```
